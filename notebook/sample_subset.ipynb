```python
import json
import random
```

```python
training_set = json.load(open("../local_labels/train_filtered.json"))
```

```python
grpo_valid = []
valid_set = json.load(open("../local_labels/val.json"))
group_cnt = {'CW': 0, 'CH': 0, 'TN': 0, 'TC': 0, 'DC': 0, 'DL': 0, 'DO': 0, 'TP': 0}
random.shuffle(valid_set)
for item in valid_set:
    # print(item['conversations'][0]['value'])
    linewise_content = item['conversations'][0]['value'].split('\n')
    # print(linewise_content)
    for i, content in enumerate(linewise_content[2:]):
        # print(content)
        linewise_content[2+i] = f"{i}, "+content
    # print(linewise_content)
    linewise_content = "\n".join(linewise_content)
    # print(linewise_content)
    item['conversations'][0]['value'] = linewise_content
    # grpo_valid.append(item)
    # print(item)
    qtype = item['id'].split('_')[0]
    # if group_cnt[qtype] > 9:
    #     continue
    # else:
    #     group_cnt[qtype] += 1
    grpo_valid.append(item)

with open("../local_labels/grpo_val.json", "w") as f:
    json.dump(grpo_valid, f, indent=2)
```

```python
grpo_pretrain = []
random.shuffle(training_set)
group_cnt = {'CW': 0, 'CH': 0, 'TN': 0, 'TC': 0, 'DC': 0, 'DL': 0, 'DO': 0, 'TP': 0}
for item in training_set:
    # print(item['conversations'][0]['value'])
    linewise_content = item['conversations'][0]['value'].split('\n')
    # print(linewise_content)
    for i, content in enumerate(linewise_content[2:]):
        # print(content)
        linewise_content[2+i] = f"{i}, "+content
    # print(linewise_content)
    linewise_content = "\n".join(linewise_content)
    # print(linewise_content)
    item['conversations'][0]['value'] = linewise_content
    qtype = item['id'].split('_')[0]
    if group_cnt[qtype] > 250:
        continue
    else:
        group_cnt[qtype] += 1
        grpo_pretrain.append(item)
    # grpo_pretrain.append(item)
print(len(grpo_pretrain))
with open("../local_labels/train_subset.json", "w") as f:
    json.dump(grpo_pretrain, f, indent=2)
```
