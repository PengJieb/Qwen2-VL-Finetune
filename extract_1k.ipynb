```python
def sample_images(folder_path, target_path, sample_count=10):

    # 确保文件夹路径有效
    folder_path = pathlib.Path(folder_path).resolve()
    if not folder_path.exists() or not folder_path.is_dir():
        print(f"错误：文件夹 {folder_path} 不存在或不是有效目录")
        return
    
    # 获取所有图像文件（支持常见图片格式）
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif']
    image_files = []
    for ext in image_extensions:
        image_files.extend(list(folder_path.glob(f"*{ext}")))
    
    # 检查是否有图像文件
    if not image_files:
        print(f"警告：文件夹 {folder_path} 中未找到图像文件")
        return
    
    # 按文件名排序（假设文件名按时间顺序命名）
    image_files.sort()
    total_count = len(image_files)
    
    if total_count <= sample_count:
        print(f"警告：图像数量({total_count})小于等于采样数量({sample_count})，不执行删除操作")
        return
    
    # 计算等间隔索引
    indices = [int(i * (total_count - 1) / (sample_count - 1)) for i in range(sample_count)]
    indices = sorted(list(set(indices)))  # 去重并排序
    
    # 保留采样图像，删除其余图像
    saved_files = []
    deleted_files = []
    
    for i, file in enumerate(image_files):
        if i in indices:
            saved_files.append(file.name)
    
    for img_name in saved_files:
        shutil.copy2(folder_path.joinpath(img_name), target_path)

    # print(saved_files)
```

```python
import os
def sample_images(folder_path, sample_count=10):

    # 确保文件夹路径有效
    folder_path = pathlib.Path(folder_path).resolve()
    if not folder_path.exists() or not folder_path.is_dir():
        print(f"错误：文件夹 {folder_path} 不存在或不是有效目录")
        return
    
    # 获取所有图像文件（支持常见图片格式）
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif']
    image_files = []
    for ext in image_extensions:
        image_files.extend(list(folder_path.glob(f"*{ext}")))
    
    # 检查是否有图像文件
    if not image_files:
        print(f"警告：文件夹 {folder_path} 中未找到图像文件")
        return
    
    # 按文件名排序（假设文件名按时间顺序命名）
    image_files.sort()
    total_count = len(image_files)
    
    if total_count <= sample_count:
        print(f"警告：图像数量({total_count})小于等于采样数量({sample_count})，不执行删除操作")
        return
    
    # 计算等间隔索引
    indices = [int(i * (total_count - 1) / (sample_count - 1)) for i in range(sample_count)]
    indices = sorted(list(set(indices)))  # 去重并排序
    
    # 保留采样图像，删除其余图像
    saved_files = []
    deleted_files = []
    
    for i, file in enumerate(image_files):
        if i in indices:
            saved_files.append(file.name)
        else:
            try:
                os.remove(file)
                deleted_files.append(file.name)
            except Exception as e:
                print(f"删除文件 {file.name} 时出错: {e}")
```

```python
def move_files_and_remove_dir(source_dir):
    # Get the parent directory
    parent_dir = os.path.dirname(source_dir)
    
    # Ensure source directory exists
    if not os.path.isdir(source_dir):
        print(f"Error: {source_dir} is not a valid directory")
        return False
    
    # Move all files from source to parent
    moved_files = []
    for item in os.listdir(source_dir):
        src_path = os.path.join(source_dir, item)
        dst_path = os.path.join(parent_dir, item)
        
        # Skip . and ..
        if item in ['.', '..']:
            continue
            
        try:
            shutil.move(src_path, dst_path)
            moved_files.append(item)
        except Exception as e:
            print(f"Error moving {item}: {e}")
    
    # Remove the original directory
    try:
        os.rmdir(source_dir)
        print(f"Removed directory: {source_dir}")
    except OSError as e:
        print(f"Error removing directory: {e}")
        return False
    
    print(f"Moved {len(moved_files)} files to {parent_dir}")
    return True
```

```python
qwenvl_train = []
droot = pathlib.Path("nextqa")
for item in tqdm(training_json):
    qid = item['qid']
    image = [str(item) for item in droot.joinpath('nextqa_frame').joinpath(item['video']).iterdir()]
    image.sort()
    depth = [str(item) for item in droot.joinpath('nextqa_depth').joinpath(item['video']).iterdir()]
    depth.sort()
    flow = [str(item) for item in droot.joinpath('nextqa_flow').joinpath(item['video']).iterdir()]
    flow.sort()
    norm = [str(item) for item in droot.joinpath('nextqa_norm').joinpath(item['video']).iterdir()]
    norm.sort()
    
    tmp_image= []
    for img in image:
        try:
            with open(img, 'r') as f:
                content = f.read(1)
                tmp_image.append(item)
        except Exception as e:
            continue
    image = tmp_image
    
    tmp_image= []
    for img in depth:
        try:
            with open(img, 'r') as f:
                content = f.read(1)
                tmp_image.append(item)
        except Exception as e:
            continue
    depth = tmp_image
    
    tmp_image= []
    for img in flow:
        try:
            with open(img, 'r') as f:
                content = f.read(1)
                tmp_image.append(item)
        except Exception as e:
            continue
    flow = tmp_image
    
    tmp_image= []
    for img in norm:
        try:
            with open(img, 'r') as f:
                content = f.read(1)
                tmp_image.append(item)
        except Exception as e:
            continue
    norm = tmp_image
    
    answer_options = [item[f"a{i}"] for i in range(item['num_option'])]
    answer_options = [f"a{i}: {answer_options[i]}" for i in range(item['num_option'])]
    answer_options_str = '\n'.join(answer_options)
    # print(answer_options)
    conversations = [
        {
            "from": 'human',
            'value': f"<image>\n{item['question']} Choose one of the following options: \n{answer_options_str}"
        },
        {
            "from": 'gpt',
            'value': f"{item['answer']}, {answer_options[item['answer']]}"
        }
    ]
    qwenvl_train.append(
        {
            "id": qid,
            "image": image,
            "depth": depth,
            "flow": flow,
            "norm": norm,
            "conversations": conversations
        }
    )
    
```
